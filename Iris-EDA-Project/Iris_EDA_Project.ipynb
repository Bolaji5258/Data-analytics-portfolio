# Iris Dataset Exploratory Data Analysis (EDA)

## Objective
Analyze the Iris dataset to understand species distribution, feature relationships, and key statistical patterns.

## Dataset
- 150 observations
- 5 columns
- 3 species:
  - Setosa
  - Versicolor
  - Virginica

## Key Findings
- No missing values were found.
- 3 duplicate records were removed.
- Petal length and petal width have a very strong positive correlation.
- Setosa has significantly smaller petal measurements than the other species.

## Conclusion

The Iris dataset was successfully analyzed using exploratory data analysis techniques. 
The dataset contained no missing values, and 3 duplicate records were removed. 
Petal length and petal width showed a strong positive correlation. 
Setosa flowers had noticeably smaller petal measurements compared to Versicolor and Virginica.

```python
import matplotlib.pyplot as plt
print("Matplotlib is working!")
```

```python
df['species'].value_counts().plot(kind='bar')

plt.title('Count of Iris Species')
plt.xlabel('Species')
plt.ylabel('Count')
plt.show()
```

```python
df['petal_length'].hist(bins=10)

plt.title('Distribution of petal length')
plt.xlabel('petal length')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.scatter(df['petal_length'], df['petal_width'])

plt.title('Petal Length vs Petal Width')
plt.xlabel('Petal Length')
plt.ylabel('Petal Width')
plt.show()
```
